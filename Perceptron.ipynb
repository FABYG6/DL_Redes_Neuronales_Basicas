```python
import numpy as np

print("NumPy esta listo")
```

# Crear matriz de datos

```python
X = np.array([
    [0.80, 0.60, 0.70],
    [0.30, 0.40, 0.20],
    [0.90, 0.80, 0.90],
    [0.40, 0.30, 0.50]
])

print(X)
print("Forma de X:", X.shape)
```

#Definir Peso de la matriz

```python
W = np.array([0.5, 0.3, 0.2])
bias = -0.5

print("Pesos:", W)
print("Bias:", bias)
```

#Calculo de la Matriz

```python
Z = X @ W + bias

print("Resultado antes de decidir:")
print(Z)
```

#Activación


```python
def funcion_escalon(z):
    return np.where(z >= 0, 1, 0)

prediccion = funcion_escalon(Z)

print("Predicciones:")
print(prediccion)
```

#vectorización

```python
Z = X @ W + bias
```

#peso para tres neuronas

```python
W1 = np.array([
    [0.5, 0.2, 0.4],
    [0.3, 0.6, 0.1],
    [0.2, 0.4, 0.7]
])

b1 = np.array([0.1, 0.1, 0.1])

Z1 = X @ W1 + b1
print(Z1)
print("Forma:", Z1.shape)
```

```python
def relu(z):
    return np.maximum(0, z)

A1 = relu(Z1)
print(A1)
```

```python
W2 = np.array([
    [0.4],
    [0.6],
    [0.5]
])

b2 = np.array([0.1])

Z2 = A1 @ W2 + b2
print(Z2)
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

A2 = sigmoid(Z2)
print("Valores entre 0 y 1:")
print(A2)

prediccion_final = np.where(A2 >= 0.5, 1, 0)
print("Prediccion final:")
print(prediccion_final)
```
