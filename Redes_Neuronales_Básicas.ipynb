```python
import numpy as np

# =======================================
# Implementación básica de un perceptrón.
# =======================================

# Datos de entrada (XOR simplificado)
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([0,0,0,1])   # salida tipo AND

# Inicialización de pesos y sesgo
np.random.seed(42)
W = np.random.randn(2)    # dos entradas
b = np.random.randn()

# Función de activación (step)
def step(x):
    return np.where(x>=0, 1, 0)

# Forward
def perceptron(X, W, b):
    return step(np.dot(X, W) + b)

print("Predicciones:", perceptron(X, W, b))
print("Forma de X:", X.shape)
```

```python
 # =========================
 # Red neuronal de una capa
 # =========================

# Entradas
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([[0],[1],[1],[0]])   # XOR

# Inicialización
np.random.seed(42)
W = np.random.randn(2, 2)   # 2 entradas → 2 neuronas
b = np.random.randn(2)

# Activación sigmoide
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Forward
def single_layer(X, W, b):
    return sigmoid(np.dot(X, W) + b)

print("Salida capa única:\n", single_layer(X, W, b))
print("Forma de X:", X.shape)
```

```python
# ================================
# Red neuronal multicapa (Vanilla)
# ================================

# Datos XOR
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([[0],[1],[1],[0]])

# Inicialización
np.random.seed(42)
W1 = np.random.randn(2, 2)   # capa oculta: 2 neuronas
b1 = np.random.randn(2)
W2 = np.random.randn(2, 1)   # capa salida: 1 neurona
b2 = np.random.randn(1)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Forward
def mlp(X, W1, b1, W2, b2):
    hidden = sigmoid(np.dot(X, W1) + b1)   # capa oculta
    output = sigmoid(np.dot(hidden, W2) + b2)  # capa salida
    return output

print("Salida MLP:\n", mlp(X, W1, b1, W2, b2))
print("Forma de X:", X.shape)
```
